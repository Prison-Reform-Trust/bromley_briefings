Testing FigureWidget and on_click functionality to try and make this visually less busy.

```python
import plotly.graph_objs as go
import random
import plotly.io as pio

f = go.FigureWidget()
f.layout.hovermode = 'closest'
f.layout.hoverdistance = -1  # Ensures no "gaps" for selecting sparse data

highlighted_opacity = 1.0
default_opacity = 0.2
default_target_index = 2  # Change this to set a different default highlighted trace

# just some traces with random data points  
num_of_traces = 5
random.seed = 42
for i in range(num_of_traces):
    y = [random.random() + i / 2 for _ in range(100)]
    trace = go.Scatter(
        y=y, 
        mode='lines', 
        line={'width': 2}, 
        opacity=highlighted_opacity if i == default_target_index else default_opacity
    )
    f.add_trace(trace)

# Custom hover event handler
def update_trace(trace, points, selector):
    if len(points.point_inds) == 0:
        return
    
    for i, _ in enumerate(f.data):
        f.data[i]['opacity'] = highlighted_opacity if i == points.trace_index else default_opacity

# Attach click event to all traces
for i in range(len(f.data)):
    f.data[i].on_click(update_trace)

# let's show the figure 
f
```
